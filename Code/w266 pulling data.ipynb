# Gathering Plant Descriptions from the Perenul Database

```python
import requests
import json
```

```python
def pull_an_plant_description (id, API_Key):
    url = f"https://perenual.com/api/v2/species/details/{id}?key={API_Key}"

    payload = {}
    headers = {}

    response = requests.request("GET", url, headers=headers, data=payload)

    with open("perenual_data.txt", "a") as f:
        f.write(response.text + ",\n")

    print(response.text)
```

```python
starting_id = 601 # Change this to your starting id @@@@@@@@@@@@@@@@@@@@@
number_of_plants_to_grab = 100 # Change to 100 @@@@@@@@@@@@@@@@@@@@@@@@@
for i in range(starting_id, number_of_plants_to_grab + starting_id):
    pull_an_plant_description(i,'sk-3DqY690e824db4c4913364') # sk-5VGK690c06c3ad94e13328, sk-3DqY690e824db4c4913364
```
