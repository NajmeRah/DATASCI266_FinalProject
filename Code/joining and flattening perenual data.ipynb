# Joining the three data sources

```python
import json
```

```python
temp_str = ''
```

```python
with open("perenual_data_1to1000.txt", "r") as f:
    temp_str += f.read()
```

```python
with open("perenual_data_1000to2000.txt", "r") as f:
    temp_str += f.read()
```

```python
with open("perenual_data_2000to3000.txt", "r") as f:
    temp_str += f.read()
```

```python
j_str = '[' + temp_str + ']'
```

```python
with open("perenual_data.json", "w") as f:
    f.write(j_str)
```

```python
with open('perenual_data.json', 'r') as file:
    json_data = json.load(file)
```
